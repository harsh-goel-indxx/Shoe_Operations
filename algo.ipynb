```python
df["datetime_date"] = pd.to_datetime(df['order_date'], format='%Y-%m-%d')
df.head()
```

```python
max_pairs_possible = 1000
```

```python
recent_orders = df.sort_values(by='datetime_date', ascending=True).head(10)
recent_orders
```

```python
recent_orders_date = recent_orders['datetime_date'].unique().tolist()
recent_orders_date
```

```python
total_count = 0
start = 0
latest_date = recent_orders_date[start]
while total_count < max_pairs_possible and start < len(recent_orders_date)-1:
    current_orders = df[df['datetime_date'] == latest_date]
    # print(current_orders)
    total_count += current_orders["quantity_pairs"].sum()
    print(total_count)
    start += 1
    latest_date = recent_orders_date[start]

    
```
